```python
import hopsworks
import pandas as pd
import requests

# 1. Open-Meteo Free API se AQI Data fetch karein (2024-2026)
url = "https://air-quality-api.open-meteo.com/v1/air-quality"
params = {
    "latitude": 31.5204,  # Lahore coordinates
    "longitude": 74.3587,
    "hourly": ["pm10", "pm2_5", "dust", "carbon_monoxide"],
    "start_date": "2024-01-01",
    "end_date": "2026-08-01",
}

response = requests.get(url, params=params)
data = response.json()

# 2. Data Frame Tayar Karein
df = pd.DataFrame(data["hourly"])
df["city"] = "Lahore"
df["date"] = pd.to_datetime(df["time"])
df["date"] = df["date"].astype("datetime64[ms]")  # Timestamp format fix
df = df.drop(columns=["time"])

# 3. Hopsworks Feature Store se Connect Karein
project = hopsworks.login()
fs = project.get_feature_store()

# 4. Feature Group Create Karein
aqi_fg = fs.get_or_create_feature_group(
    name="aqi_features",
    version=1,
    primary_key=["city", "date"],
    event_time="date",
    description="AQI historical feature group from Open-Meteo API",
)

# Data Upload Karein
aqi_fg.insert(df)
print("Data successfully uploaded to Hopsworks Feature Group!")
```

```python
# 1. Feature Group get karein
aqi_fg = fs.get_feature_group(name="aqi_features", version=1)

# 2. Query define karein
selected_features = aqi_fg.select_all()

# 3. Feature View create karein
feature_view = fs.get_or_create_feature_view(
    name="aqi_feature_view",
    version=1,
    labels=["pm2_5"],  # Target variable jo hum predict karna chahte hain
    query=selected_features,
)

print("Feature View successfully created!")
```

```python
# ============================================================
# AQI PREDICTION - HOPSWORKS DATA → XGBOOST
# Hyderabad / Lahore, Pakistan
# ============================================================

import numpy as np
import pandas as pd
import xgboost as xgb
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ------------------------------------------------------------
# 1. Make copies
# ------------------------------------------------------------
X_train = X_train.copy()
X_test = X_test.copy()
y_train = y_train.copy()
y_test = y_test.copy()

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

# ------------------------------------------------------------
# 2. Extract Date Features & Handle Object Columns
# ------------------------------------------------------------

# Extract Date components
for df in [X_train, X_test]:
    for col in list(df.columns):
        if "date" in col.lower() or "time" in col.lower():
            df[col] = pd.to_datetime(df[col], errors="coerce")
            df["hour"] = df[col].dt.hour
            df["day"] = df[col].dt.day
            df["month"] = df[col].dt.month
            df["dayofweek"] = df[col].dt.dayofweek
            df.drop(columns=[col], inplace=True)

# Categorical columns handling
object_cols = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

for col in object_cols:
    X_train[col] = X_train[col].astype("category")
    X_test[col] = X_test[col].astype("category")

    categories = X_train[col].cat.categories
    X_test[col] = pd.Categorical(X_test[col], categories=categories)


# ------------------------------------------------------------
# 3. Convert boolean columns if present
# ------------------------------------------------------------

bool_cols = X_train.select_dtypes(include=["bool"]).columns

for col in bool_cols:
    X_train[col] = X_train[col].astype(int)
    X_test[col] = X_test[col].astype(int)


# ------------------------------------------------------------
# 4. Convert target to 1D Series and numeric
# ------------------------------------------------------------

if isinstance(y_train, pd.DataFrame):
    y_train = y_train.squeeze()

if isinstance(y_test, pd.DataFrame):
    y_test = y_test.squeeze()

y_train = pd.to_numeric(y_train, errors="coerce")
y_test = pd.to_numeric(y_test, errors="coerce")


# ------------------------------------------------------------
# 5. Remove rows having invalid target values
# ------------------------------------------------------------

train_valid = y_train.notna()
test_valid = y_test.notna()

X_train = X_train.loc[train_valid].reset_index(drop=True)
y_train = y_train.loc[train_valid].reset_index(drop=True)

X_test = X_test.loc[test_valid].reset_index(drop=True)
y_test = y_test.loc[test_valid].reset_index(drop=True)


# ------------------------------------------------------------
# 6. Train XGBoost Model
# ------------------------------------------------------------

model = xgb.XGBRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=6,
    random_state=42,
    objective="reg:squarederror",
    enable_categorical=True,
)

model.fit(X_train, y_train)


# ------------------------------------------------------------
# 7. Predict & Evaluate
# ------------------------------------------------------------

preds = model.predict(X_test)

mae = mean_absolute_error(y_test, preds)
rmse = np.sqrt(mean_squared_error(y_test, preds))
r2 = r2_score(y_test, preds)

print("\n========================================")
print("      AQI MODEL EVALUATION")
print("========================================")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
print("========================================")


# ------------------------------------------------------------
# 8. Save Model to Hopsworks Model Registry
# ------------------------------------------------------------

mr = project.get_model_registry()

hopsworks_model = mr.python.create_model(
    name="aqi_prediction_model",
    metrics={"rmse": float(rmse), "mae": float(mae), "r2": float(r2)},
    description="XGBoost model for predicting PM2.5 AQI levels",
)

# Local dir mein save karke Registry mein upload karein
model.save_model("model.json")
hopsworks_model.save("model.json")

print("\nModel successfully uploaded to Hopsworks Model Registry!")
```

```python
import hopsworks
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import requests
import xgboost as xgb

# 1. Hopsworks se Model Load karein
project = hopsworks.login()
mr = project.get_model_registry()
model_meta = mr.get_model("aqi_prediction_model", version=1)
model_dir = model_meta.download()

model = xgb.XGBRegressor()
model.load_model(f"{model_dir}/model.json")

# 2. Open-Meteo se Next 3 Days Forecast Data Fetch karein
forecast_url = "https://air-quality-api.open-meteo.com/v1/air-quality"
params = {
    "latitude": 31.5204,
    "longitude": 74.3587,
    "hourly": ["pm10", "dust", "carbon_monoxide"],
    "forecast_days": 3,
}
res = requests.get(forecast_url, params=params).json()

forecast_df = pd.DataFrame(res["hourly"])
forecast_df["date"] = pd.to_datetime(forecast_df["time"])

# Columns matching exact training schema
features = forecast_df[["pm10", "dust", "carbon_monoxide"]]

# 3. 3-Day AQI Predictions
forecast_df["predicted_pm2_5"] = model.predict(features)


# Alert status function
def get_alert_level(pm25):
    if pm25 <= 35:
        return "Good"
    elif pm25 <= 75:
        return "Moderate"
    elif pm25 <= 150:
        return "Unhealthy for Sensitive Groups"
    else:
        return "HAZARDOUS 🚨"


forecast_df["alert_level"] = forecast_df["predicted_pm2_5"].apply(
    get_alert_level
)
forecast_df.to_csv("aqi_3day_forecast.csv", index=False)
print("3-Day Predictions successfully generated and saved!")

# 4. Feature Importance Plot
plt.figure(figsize=(6, 4))
xgb.plot_importance(model, importance_type="weight")
plt.title("XGBoost Feature Importance")
plt.tight_layout()
plt.savefig("feature_importance.png")
print("Feature Importance plot saved as 'feature_importance.png'!")
```
