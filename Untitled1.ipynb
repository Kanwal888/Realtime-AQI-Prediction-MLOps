```python
%%writefile app.py
import pandas as pd
import plotly.express as px
import streamlit as st

# Page Configuration
st.set_page_config(page_title="AQI Forecast App", page_icon="🌬️", layout="wide")

st.title("🌬️ Real-Time AQI Prediction & Analytics Web App")
st.caption("Powered by Hopsworks Feature Store & XGBoost Engine")

try:
    df = pd.read_csv("aqi_3day_forecast.csv")
    current = df.iloc[0]

    # Metric Cards
    c1, c2, c3, c4 = st.columns(4)
    c1.metric("Predicted PM2.5", f"{current['predicted_pm2_5']:.2f} µg/m³")
    c2.metric("PM10", f"{current['pm10']} µg/m³")
    c3.metric("Dust Level", f"{current['dust']} µg/m³")
    c4.metric("Alert Status", current["alert_level"])

    if "HAZARDOUS" in str(current["alert_level"]):
        st.error("🚨 **HAZARDOUS ALERT**: High air pollution! Wear masks.")

    st.markdown("---")

    # Interactive Plotly Line Chart
    st.subheader("📈 3-Day Hourly AQI Forecast Trend")
    fig = px.line(
        df, 
        x="date", 
        y="predicted_pm2_5", 
        labels={"date": "Date & Time", "predicted_pm2_5": "PM2.5 (µg/m³)"},
        color_discrete_sequence=["#EF553B"]
    )
    fig.update_traces(mode="lines+markers")
    st.plotly_chart(fig, use_container_width=True)

    # Split View
    col_l, col_r = st.columns(2)
    with col_l:
        st.subheader("🔍 Feature Importance")
        st.image("feature_importance.png", use_container_width=True)

    with col_r:
        st.subheader("📊 Forecast Dataset Table")
        st.dataframe(df[["date", "predicted_pm2_5", "pm10", "dust", "alert_level"]], height=350)

except Exception as e:
    st.error("Pehle Inference script run karke data generate karein.")
```

```python
import gradio as gr
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# Data Load & Processing Function
def load_dashboard_data():
    df = pd.read_csv("aqi_3day_forecast.csv")
    current = df.iloc[0]

    pm25 = f"{current['predicted_pm2_5']:.2f} µg/m³"
    pm10 = f"{current['pm10']} µg/m³"
    dust = f"{current['dust']} µg/m³"
    alert = current["alert_level"]

    # 1. Trend Plot (Matplotlib Dark Theme)
    fig, ax = plt.subplots(figsize=(10, 4), facecolor="#0f172a")
    ax.set_facecolor("#1e293b")

    times = pd.to_datetime(df["date"])
    ax.plot(
        times,
        df["predicted_pm2_5"],
        color="#38bdf8",
        linewidth=2.5,
        label="Predicted PM2.5",
    )
    ax.axhline(
        y=150,
        color="#ef4444",
        linestyle="--",
        linewidth=1.5,
        label="Hazardous Threshold (150)",
    )

    ax.tick_params(colors="#94a3b8")
    ax.spines["bottom"].set_color("#334155")
    ax.spines["top"].set_color("#334155")
    ax.spines["left"].set_color("#334155")
    ax.spines["right"].set_color("#334155")
    ax.set_ylabel("PM2.5 (µg/m³)", color="#f8fafc")
    ax.grid(True, color="#334155", alpha=0.5)
    ax.legend(facecolor="#1e293b", edgecolor="#334155", labelcolor="#f8fafc")
    plt.tight_layout()

    # 2. Feature Importance Plot
    fig_feat, ax_feat = plt.subplots(figsize=(6, 4), facecolor="#0f172a")
    ax_feat.set_facecolor("#1e293b")
    try:
        feat_img = plt.imread("feature_importance.png")
        ax_feat.imshow(feat_img)
        ax_feat.axis("off")
    except:
        pass

    # 3. Data Table
    table_data = df[
        ["date", "predicted_pm2_5", "pm10", "dust", "alert_level"]
    ].head(10)

    return pm25, pm10, dust, alert, fig, fig_feat, table_data


# Build Gradio UI
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("""
    # 🌬️ Air Quality Forecasting & Analytics Dashboard
    ### Real-Time AQI Predictions powered by Hopsworks & XGBoost
    """)

    with gr.Row():
        val_pm25 = gr.Textbox(label="Predicted PM2.5", interactive=False)
        val_pm10 = gr.Textbox(label="PM10 Level", interactive=False)
        val_dust = gr.Textbox(label="Dust Level", interactive=False)
        val_alert = gr.Textbox(label="Air Quality Status", interactive=False)

    gr.Markdown("---")

    with gr.Row():
        plot_trend = gr.Plot(label="📈 3-Day Hourly AQI Forecast Trend")

    with gr.Row():
        plot_feat = gr.Plot(label="🔍 Feature Importance")
        table_forecast = gr.Dataframe(label="📊 Forecast Data Sample")

    demo.load(
        load_dashboard_data,
        outputs=[
            val_pm25,
            val_pm10,
            val_dust,
            val_alert,
            plot_trend,
            plot_feat,
            table_forecast,
        ],
    )

# Launch Dashboard
demo.launch(share=True, inline=True)
```
